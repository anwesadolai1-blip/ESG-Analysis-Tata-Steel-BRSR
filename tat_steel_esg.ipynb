```python
def plot_metric(column, ylabel):
    plt.figure(figsize=(8,5))
    plt.plot(df["Year"], df[column], marker="o", linewidth=2)
    plt.title(column)
    plt.xlabel("Financial Year")
    plt.ylabel(ylabel)
    plt.grid(True)
    plt.show()
```

```python
plot_metric("EnergyConsumed", "PJ")
plot_metric("Scope1", "Mt CO₂e")
plot_metric("Scope2", "Mt CO₂e")
plot_metric("EmissionIntensity", "tCO₂e/t crude steel")
plot_metric("WaterWithdrawal", "Million Litres")
plot_metric("WaterConsumption", "Million Litres")
plot_metric("WaterIntensity", "kL/t crude steel")
```
